```python
def fetch_video_data(video_id):
    try:
        ytt = YouTubeTranscriptApi()
        fetched_transcript = ytt.fetch(video_id)

        video_data = []
        for snippet in fetched_transcript:
            video_data.append({
                "text": snippet.text,
                "start": snippet.start,
                "duration": snippet.duration
            })
        
        return video_data
    except Exception as e:
        raise
```

```python
video_data = fetch_video_data(video_id=video_id)
```

```python
import pickle

def load_chunks(file_path:str = "data/chunks.pkl") :
        """
        Loads chunks from pkl file. Chunks will be a list of dicts containing chunk text, start and end time
        """
        with open(file_path, "rb") as f:
            chunks = pickle.load(f)
        return chunks

        
```

```python
chunks = load_chunks(file_path= "../data/chunks.pkl")
chunks[0]['text']
```

```python
for i, chunk in enumerate(chunks):
    if 'start' not in chunk:
        print("index: ", i)
        print(chunk)
```

```python
texts = [item['text'] for item in chunks]
```
